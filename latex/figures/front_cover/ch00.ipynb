```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def plot_saddle_surface(
    range_xy=(-2, 2),   # XY轴的范围，(min, max)
    resolution=600,     # 网格分辨率，分辨率越高越细腻
    cmap='coolwarm',    # 颜色映射方案
    output_file='output.pdf',   # 导出pdf文件
    save=True,          # 是否保存
    show=True           # 是否显示
):
    """
    生成极高分辨率的马鞍面深度学习封面图。
    Saddle point 象征着梯度下降面临的局部优化难题。
    """    
    # 生成高密度网格数据
    x = np.linspace(range_xy[0], range_xy[1], resolution)
    y = np.linspace(range_xy[0], range_xy[1], resolution)
    x, y = np.meshgrid(x, y)
    
    # 马鞍面公式 z = x^2 - y^2
    z = x**2 - y**2

    # 创建白色背景图像
    # 将 fig 的 dpi 也设为 1000，保证渲染时的精度与导出时一致
    fig = plt.figure(figsize=(10, 10), dpi=1000, facecolor='white')
    ax = fig.add_subplot(111, projection='3d')

    # 绘制马鞍面，使用红蓝渐变，去除网格线和边框
    surf = ax.plot_surface(
        x, y, z, 
        cmap=cmap, 
        edgecolor='none', 
        antialiased=True, 
        shade=True,
    )

    # 彻底去除坐标轴和边框
    ax.set_axis_off()

    # 紧凑布局，最大化几何图形展示
    plt.subplots_adjust(left=0, right=1, top=1, bottom=0)

    # 保存与显示
    if save:
        plt.savefig(
            output_file, 
            format='pdf',           # 显式指定输出格式为 pdf 矢量图
            dpi=1000,               # 设置dpi=1000
            bbox_inches='tight',    # 彻底裁掉边缘空白
            pad_inches=0,
            transparent=True        # 背景透明，方便排版时叠加
        )
        print(f"图像已保存至: {output_file}")
    
    if show:
        plt.show()
    
    plt.close(fig) # 释放内存

# 运行生成
plot_saddle_surface(
    resolution=600, 
    cmap='coolwarm', 
    output_file='./cover.pdf'
)
```
